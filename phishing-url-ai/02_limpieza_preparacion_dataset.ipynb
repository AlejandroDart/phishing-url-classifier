```python
# ============================================================
# Celda 1: Importación de librerías y configuración inicial
# ============================================================
# En esta celda se importan las librerías necesarias para la limpieza
# y preparación del dataset. También se define la ruta del archivo original,
# la variable objetivo y la paleta visual base.

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)

DATASET_PATH = "dataset-phishing.csv"
TARGET_COL = "status"

COLOR_PRIMARIO = "#3A6EA5"
COLOR_SECUNDARIO = "#5DADE2"
COLOR_TERCIARIO = "#95A5A6"

sns.set_theme(style="whitegrid")
```

```python
# ============================================================
# Celda 2: Carga del dataset original
# ============================================================
# En esta celda se carga el dataset original y se genera una copia de trabajo.
# Esto permite mantener intacto el archivo base durante todo el proceso.

df_original = pd.read_csv(DATASET_PATH)

df_clean = df_original.copy()

print("Shape original:", df_original.shape)
print("Shape copia de trabajo:", df_clean.shape)

display(df_clean.head())
```

```python
# ============================================================
# Celda 3: Conversión de la variable objetivo a formato binario
# ============================================================
# En esta celda se convierte la variable objetivo a formato binario,
# donde:
# 0 = legítima
# 1 = phishing
#
# El dataset ya contiene únicamente las clases 'legitimate' y 'phishing',
# por lo que se realiza un mapeo directo.

df_clean[TARGET_COL] = (
    df_clean[TARGET_COL]
    .astype(str)
    .str.lower()
    .str.strip()
)

df_clean[TARGET_COL] = df_clean[TARGET_COL].map({
    "legitimate": 0,
    "phishing": 1
})

df_clean = df_clean.dropna(subset=[TARGET_COL])

df_clean[TARGET_COL] = df_clean[TARGET_COL].astype(int)

print("Valores únicos del target:")
print(df_clean[TARGET_COL].unique())

print("\nDistribución de clases:")
display(df_clean[TARGET_COL].value_counts())

print("\nShape después de transformar target:")
print(df_clean.shape)
```

```python
# ============================================================
# Celda 4: Eliminación de URLs duplicadas
# ============================================================
# En esta celda se eliminan registros con URLs repetidas, conservando
# la primera aparición. Esto evita que una misma URL aparezca más de una vez
# durante el entrenamiento o evaluación del modelo.

duplicadas_url = df_clean["url"].duplicated().sum()

print("URLs duplicadas antes de limpiar:", duplicadas_url)

if duplicadas_url > 0:
    urls_duplicadas = df_clean[df_clean["url"].duplicated(keep=False)].sort_values("url")
    print("\nRegistros duplicados detectados:")
    display(urls_duplicadas)

df_clean = df_clean.drop_duplicates(subset=["url"], keep="first")

print("\nURLs duplicadas después de limpiar:", df_clean["url"].duplicated().sum())
print("Shape después de eliminar URLs duplicadas:", df_clean.shape)
```

```python
# ============================================================
# Celda 5: Eliminación de filas duplicadas completas
# ============================================================
# En esta celda se eliminan filas completamente duplicadas.
# Aunque en el EDA no se observaron duplicados completos, se mantiene
# esta validación para asegurar consistencia en el pipeline.

duplicadas_filas = df_clean.duplicated().sum()

print("Filas duplicadas completas antes de limpiar:", duplicadas_filas)

df_clean = df_clean.drop_duplicates(keep="first")

print("Filas duplicadas completas después de limpiar:", df_clean.duplicated().sum())
print("Shape después de eliminar filas duplicadas:", df_clean.shape)
```

```python
# ============================================================
# Celda 6: Identificación y eliminación de variables constantes
# ============================================================
# En esta celda se detectan variables que poseen un único valor en todo
# el dataset. Estas variables no aportan información predictiva y pueden
# eliminarse antes del entrenamiento.

unique_counts = df_clean.nunique()

constant_features = unique_counts[unique_counts <= 1].index.tolist()

print("Variables constantes detectadas:")
print(constant_features)

print("\nCantidad de variables constantes:", len(constant_features))

df_clean = df_clean.drop(columns=constant_features)

print("\nShape después de eliminar variables constantes:", df_clean.shape)
```

```python
# ============================================================
# Celda 7: Revisión y tratamiento de valores nulos
# ============================================================
# En esta celda se revisan valores nulos. Si existen, se imputan usando
# la mediana en variables numéricas. Esto evita errores posteriores
# durante el entrenamiento de modelos.

missing_before = df_clean.isnull().sum()
missing_before = missing_before[missing_before > 0].sort_values(ascending=False)

print("Valores nulos antes de imputar:")
display(missing_before)

numeric_cols = df_clean.select_dtypes(include=["int64", "float64"]).columns

for col in numeric_cols:
    if df_clean[col].isnull().sum() > 0:
        df_clean[col] = df_clean[col].fillna(df_clean[col].median())

missing_after = df_clean.isnull().sum()
missing_after = missing_after[missing_after > 0].sort_values(ascending=False)

print("\nValores nulos después de imputar:")
display(missing_after)

if missing_after.empty:
    print("No quedan valores nulos en el dataset.")
```

```python
# ============================================================
# Celda 8: Revisión de valores infinitos
# ============================================================
# En esta celda se reemplazan posibles valores infinitos por NaN
# y luego se imputan con la mediana de cada variable numérica.

numeric_cols = df_clean.select_dtypes(include=["int64", "float64"]).columns

inf_count = np.isinf(df_clean[numeric_cols]).sum()
inf_count = inf_count[inf_count > 0]

print("Valores infinitos detectados:")
display(inf_count)

df_clean[numeric_cols] = df_clean[numeric_cols].replace([np.inf, -np.inf], np.nan)

for col in numeric_cols:
    if df_clean[col].isnull().sum() > 0:
        df_clean[col] = df_clean[col].fillna(df_clean[col].median())

print("Valores infinitos tratados correctamente.")
```

```python
# ============================================================
# Celda 9: Revisión de valores negativos especiales
# ============================================================
# En esta celda se identifican variables numéricas que contienen
# valores negativos. En datasets de phishing, ciertos atributos
# asociados a consultas WHOIS pueden utilizar valores como -1
# para representar:
#
# - ausencia de información
# - fallos en la consulta
# - dominios ocultos o inválidos
#
# Debido a que estos valores pueden contener información relevante
# para la detección de phishing, no serán eliminados ni imputados
# en esta etapa.

numeric_cols = df_clean.select_dtypes(include=["int64", "float64"]).columns

negative_report = []

for col in numeric_cols:
    
    negative_count = (df_clean[col] < 0).sum()
    
    if negative_count > 0:
        
        negative_report.append({
            "feature": col,
            "negative_count": negative_count,
            "negative_percent": round(
                negative_count / len(df_clean) * 100,
                2
            ),
            "min_value": df_clean[col].min()
        })

negative_report_df = (
    pd.DataFrame(negative_report)
    .sort_values("negative_percent", ascending=False)
)

display(negative_report_df)

if negative_report_df.empty:
    print("No se encontraron valores negativos.")
else:
    print("""
Nota:
Los valores negativos detectados se conservarán temporalmente,
ya que podrían representar información relevante asociada
a consultas WHOIS fallidas o dominios sospechosos.
""")
```

```python
# ============================================================
# Celda 10: Conversión segura de variables numéricas
# ============================================================
# En esta celda se asegura que todas las variables predictoras,
# excepto la URL, estén en formato numérico. Cualquier valor no convertible
# se transforma en NaN y se imputa con la mediana.

cols_no_predictoras = ["url", TARGET_COL]

feature_cols = [col for col in df_clean.columns if col not in cols_no_predictoras]

for col in feature_cols:
    df_clean[col] = pd.to_numeric(df_clean[col], errors="coerce")
    df_clean[col] = df_clean[col].fillna(df_clean[col].median())

print("Tipos de datos después de conversión:")
display(df_clean.dtypes.value_counts())

print("\nShape actual:", df_clean.shape)
```

```python
# ============================================================
# Celda 11: Validación final del dataset limpio
# ============================================================
# En esta celda se realiza una validación final del dataset limpio:
# dimensiones, nulos, duplicados, tipos de datos y distribución de clases.

print("Shape final:", df_clean.shape)

print("\nValores nulos totales:", df_clean.isnull().sum().sum())

print("Filas duplicadas completas:", df_clean.duplicated().sum())

if "url" in df_clean.columns:
    print("URLs duplicadas:", df_clean["url"].duplicated().sum())

print("\nDistribución final del target:")
display(df_clean[TARGET_COL].value_counts())

print("\nTipos de datos:")
display(df_clean.dtypes.value_counts())

display(df_clean.head())
```

```python
# ============================================================
# Celda 12: Guardado del dataset limpio
# ============================================================
# En esta celda se guarda el dataset limpio generado durante la etapa
# de preparación. Este archivo será utilizado en el siguiente notebook
# para la selección de características.

OUTPUT_PATH = "dataset-phishing-clean.csv"

df_clean.to_csv(OUTPUT_PATH, index=False)

print("Dataset limpio guardado correctamente.")
print("Archivo:", OUTPUT_PATH)
print("Shape guardado:", df_clean.shape)
```
