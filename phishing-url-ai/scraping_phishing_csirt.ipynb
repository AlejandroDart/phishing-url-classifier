```python
import requests
from bs4 import BeautifulSoup
import csv

# Función para obtener las URLs de las publicaciones desde una lista de páginas de búsqueda
def obtener_urls_de_paginas(paginas_busqueda):
    all_links = []
    base_url = "https://csirt.gob.cl"
    
    for search_url in paginas_busqueda:
        response = requests.get(search_url)
        if response.status_code == 200:
            soup = BeautifulSoup(response.content, 'html.parser')
            
            # Buscar todos los elementos con la clase 'mb-1' que contienen los enlaces
            elementos = soup.find_all('a', class_='mb-1')
            
            # Extraer los href de los enlaces
            links = [elemento['href'] for elemento in elementos if 'href' in elemento.attrs]
            
            # Crear una lista de URLs completas
            full_links = [base_url + link if link.startswith("/") else link for link in links]
            all_links.extend(full_links)
        else:
            print(f"Error al acceder a la página: {search_url} (Status code: {response.status_code})")
    
    return list(set(all_links))  # Eliminar duplicados

# Función para extraer URLs desde una publicación específica
def extraer_urls_desde_publicacion(url_publicacion):
    response = requests.get(url_publicacion)
    soup = BeautifulSoup(response.content, 'html.parser')
    
    # Buscar la tabla de "Indicadores de Compromiso"
    tabla = soup.find('table')
    urls = []
    
    if tabla:
        # Buscar filas de la tabla
        filas = tabla.find_all('tr')
        for fila in filas:
            columnas = fila.find_all('td')
            if len(columnas) > 1:  # Verificar que hay columnas en la fila
                tipo = columnas[0].text.strip()  # Tipo (IPv4, URL, etc.)
                valor = columnas[1].text.strip()  # Valor (la URL, IP, etc.)
                if tipo.lower() == "url":  # Solo guardar las URLs
                    urls.append(valor.replace(" ", ""))  # Remover espacios dentro de la URL
    return urls

# Lista de páginas de búsqueda a procesar
paginas_busqueda = [
    "https://csirt.gob.cl/buscar/?q=phishing",
    "https://csirt.gob.cl/buscar/?q=phishing&p=2",
    "https://csirt.gob.cl/buscar/?q=phishing&p=3",
    "https://csirt.gob.cl/buscar/?q=phishing&p=4",
    "https://csirt.gob.cl/buscar/?q=phishing&p=5",
    "https://csirt.gob.cl/buscar/?q=phishing&p=6",
    "https://csirt.gob.cl/buscar/?q=phishing&p=7"
]

# Paso 1: Obtener enlaces de las publicaciones desde las páginas de búsqueda
publicaciones = obtener_urls_de_paginas(paginas_busqueda)

# Paso 2: Extraer URLs desde cada publicación
todas_las_urls = []
for publicacion in publicaciones:
    urls = extraer_urls_desde_publicacion(publicacion)
    todas_las_urls.extend(urls)  # Agregar todas las URLs extraídas a la lista

# Eliminar duplicados en las URLs
todas_las_urls = list(set(todas_las_urls))

# Guardar resultados en un archivo CSV
with open("urls_extraidas_phishing.csv", "w", newline='', encoding='utf-8') as file:
    writer = csv.writer(file)
    for url in todas_las_urls:
        writer.writerow([url])  # Escribir cada URL en una línea sin encabezado

print("Las URLs han sido almacenadas en 'urls_extraidas.csv'.")
```
